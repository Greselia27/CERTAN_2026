```python
#Tugas 1
import numpy as np

# 1. Membuat matriks 5x5 dengan nilai random (integer antara 1 sampai 100)
matriks = np.random.randint(1, 101, size=(5, 5))

# Menampilkan matriks yang terbentuk
print("Matriks 5x5:")
print(matriks)

# 2. Menghitung nilai maksimum, minimum, dan rata-rata
nilai_maksimum = np.max(matriks)
nilai_minimum = np.min(matriks)
nilai_rata_rata = np.mean(matriks)

# Menampilkan hasil perhitungan
print("\n--- Hasil Analisis Matriks ---")
print(f"Nilai Maksimum : {nilai_maksimum}")
print(f"Nilai Minimum  : {nilai_minimum}")
print(f"Nilai Rata-rata: {nilai_rata_rata:.2f}")
```

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Membuat rentang koordinat x dan y
x = np.linspace(-1.5, 1.5, 500)
y = np.linspace(-1.5, 1.5, 500)
X, Y = np.meshgrid(x, y)

# 2. Menuliskan persamaan kurva
F = (X**6 + 3*(X**4)*(Y**2) - 3*(X**4) + 3*(X**2)*(Y**4) 
     - (X**2)*(Y**3) - 6*(X**2)*(Y**2) + 3*(X**2) 
     + Y**6 - 3*(Y**4) + 3*(Y**2))

# 3. Menggambar grafik 2 dimensi menggunakan contour pada level 1
plt.figure(figsize=(6, 6))
plt.contour(X, Y, F, levels=[1], colors='red', linewidths=2)

# Mengatur tampilan grafik
plt.title("Grafik 2 Dimensi: Heart Curve", fontsize=12)
plt.xlabel("Sumbu X")
plt.ylabel("Sumbu Y")
plt.axhline(0, color='gray', linestyle='--', linewidth=0.7)
plt.axvline(0, color='gray', linestyle='--', linewidth=0.7)
plt.grid(True, linestyle=':', alpha=0.6)
plt.gca().set_aspect('equal', adjustable='box')

plt.show()
```

```python
#Tugas 3
import numpy as np
import matplotlib.pyplot as plt

# 1. Membuat grid koordinat untuk sumbu X dan Y
x = np.linspace(-2, 2, 200)
y = np.linspace(-2, 2, 200)
X, Y = np.meshgrid(x, y)

# 2. Menghitung nilai Z (fungsi 3 dimensi)
Z = (5 * Y / 4 - np.sqrt(np.abs(X)))**2 + X**2

# 3. Membuat grafik permukaan 3 Dimensi (3D Surface Plot)
fig = plt.figure(figsize=(9, 7))
ax = fig.add_subplot(111, projection='3d')

# Memplot permukaan dengan colormap viridis
surf = ax.plot_surface(X, Y, Z, cmap='viridis', edgecolor='none', alpha=0.9)

# Memberikan label pada sumbu
ax.set_title(r"Grafik 3 Dimensi: $z = \left(\frac{5y}{4} - \sqrt{|x|}\right)^2 + x^2$", fontsize=12)
ax.set_xlabel("Sumbu X")
ax.set_ylabel("Sumbu Y")
ax.set_zlabel("Sumbu Z")

# Menambahkan colorbar
fig.colorbar(surf, ax=ax, shrink=0.5, aspect=10)

plt.show()
```
