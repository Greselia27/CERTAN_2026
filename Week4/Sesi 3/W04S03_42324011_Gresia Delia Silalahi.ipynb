```python
import numpy as np
import pandas as pd
from sklearn.datasets import load_iris

# Load dataset Iris dari scikit-learn
iris = load_iris()
```

##### Penjelasan:
##### Sel ini mengimpor seluruh pustaka komputasi numerik dan pemrosesan tabel data yang dibutuhkan, kemudian memuat dataset bunga Iris bawaan dari pustaka pembelajaran mesin ke dalam sebuah objek siap pakai yang menyimpan data pengukuran kelopak dan mahkota beserta label spesies bunganya.

```python
# Konversi data menjadi DataFrame Pandas
df = pd.DataFrame(iris.data, columns=iris.feature_names)
df["Species"] = iris.target_names[iris.target]

# Definisikan X (fitur) dan y (target/label)
X = df.drop(columns=["Species"])
y = df["Species"]

print(X)
```

##### Penjelasan:
##### Sel ini mengubah data mentah menjadi tabel terstruktur yang rapi dengan menyertakan nama kolom untuk setiap dimensi fisik bunga, menambahkan kolom nama spesies, lalu memisahkan data menjadi sekumpulan variabel penjelas yang berisi empat karakteristik pengukuran serta variabel target yang berisi spesies bunga, kemudian menampilkan seluruh data pengukuran tersebut ke layar dalam format tabel seratus lima puluh baris dan empat kolom.

```python
y
```

##### Penjelasan:
##### Sel ini mengecek dan menampilkan isi dari variabel target yang berisi label jenis spesies untuk setiap bunga dalam dataset, menghasilkan daftar seratus lima puluh data bertipe teks yang memuat nama spesies seperti setosa, versicolor, dan virginica sesuai dengan urutan baris data masing-masing sampel.

```python
from sklearn.preprocessing import LabelEncoder
lab = LabelEncoder()
y = lab.fit_transform(y)
y
```

##### Penjelasan:
##### Sel ini mengonversi label teks nama spesies bunga menjadi angka-angka bilangan bulat agar dapat diolah secara matematis oleh algoritma klasifikasi, di mana hasil konversinya memetakan spesies setosa menjadi angka nol, versicolor menjadi angka satu, dan virginica menjadi angka dua untuk seluruh seratus lima puluh sampel data.

```python
from sklearn.model_selection import train_test_split

X_train, X_test = train_test_split(X, test_size=0.2, random_state=0)
y_train, y_test = train_test_split(y, test_size=0.2, random_state=0)
```

##### Penjelasan:
##### Sel ini membagi kumpulan data fitur dan label target menjadi dua kelompok terpisah secara proporsional, yaitu data pelatihan sebanyak delapan puluh persen untuk melatih model cerdas dan data pengujian sebanyak dua puluh persen atau tiga puluh sampel untuk menguji performa model, dengan pengacakan terkontrol agar hasil pembagian data selalu konsisten ketika dieksekusi berulang kali.

```python
from sklearn.naive_bayes import GaussianNB

model = GaussianNB()
NB_model = model.fit(X_train, y_train)
```

##### Penjelasan:
##### Sel ini membuat model pengklasifikasi berbasis probabilitas dengan asumsi sebaran normal untuk setiap ukuran fisik bunga, kemudian melatih model tersebut menggunakan sekumpulan data fitur dan label dari data pelatihan agar model mempelajari nilai rata-rata serta rentang sebaran dari masing-masing spesies bunga.

```python
result = NB_model.predict(X_test)
result
```

##### Penjelasan:
##### Sel ini menjalankan proses prediksi terhadap seluruh data pengujian yang belum pernah dipelajari sebelumnya untuk menebak spesies masing-masing bunga, dan menghasilkan luaran berupa deretan tiga puluh angka prediksi yang menunjukkan hasil tebakan kategori spesies untuk setiap sampel data uji tersebut.

```python
from sklearn.metrics import accuracy_score
print('Accuracy of data testing: ', accuracy_score(y_test, result)*100)
```

##### Penjelasan:
##### Sel ini mengukur tingkat ketepatan hasil prediksi model dengan membandingkan tebakan spesies terhadap jawaban asli pada data pengujian, menghasilkan skor akurasi sekitar sembilan puluh enam koma enam puluh tujuh persen yang menandakan bahwa model berhasil menebak dua puluh sembilan dari tiga puluh sampel data uji dengan sangat akurat dan hanya salah pada satu sampel saja.

```python
#prediksi inputan data baru
#input_data = pd.DataFrame(np.array([[5.1, 3.5, 1.4, 0.2]]), columns=iris.feature_names)
input_data = [[5.1, 3.5, 1.4, 0.2]]
result = NB_model.predict(input_data)
prediction_label = lab.inverse_transform(result)
print('Hasil prediksi untuk input baru adalah:', prediction_label)
```

##### Penjelasan:
##### Sel ini menguji kemampuan model untuk memprediksi data pengukuran bunga baru yang dimasukkan dalam bentuk daftar angka biasa tanpa menyertakan nama-nama fiturnya, di mana model sukses mengenali bahwa bunga tersebut berjenis setosa namun sistem memberikan pesan peringatan karena format masukan tidak memiliki nama kolom yang sesuai dengan struktur data saat model dilatih.

```python
input_data = pd.DataFrame(np.array([[5.1, 3.5, 1.4, 0.2]]), columns=iris.feature_names)

result = NB_model.predict(input_data)
prediction_label = lab.inverse_transform(result)
print('Hasil prediksi untuk input baru adalah:', prediction_label)
```

##### Penjelasan:
##### Sel ini memperbaiki kekurangan pada sel sebelumnya dengan mengemas data masukan baru ke dalam bentuk tabel yang lengkap dengan nama-nama kolom fitur yang identik dengan data latih, sehingga model dapat memprediksi bunga tersebut sebagai spesies setosa secara akurat dan bersih tanpa menimbulkan pesan peringatan sedikit pun.

```python
# TUGAS
# Prediksi Data Baru

input_data_tugas = pd.DataFrame(np.array([[7.3, 3.8, 4.6, 2.3]]), columns=iris.feature_names)

result_tugas = NB_model.predict(input_data_tugas)
prediction_label_tugas = lab.inverse_transform(result_tugas)
print('Hasil prediksi untuk data baru adalah:', prediction_label_tugas)
```
